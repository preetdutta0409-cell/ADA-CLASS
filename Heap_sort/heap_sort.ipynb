```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("heapsortN.csv")
```

```python
plt.figure(figsize=(12, 7))

plt.plot(
    df["ArraySize"],
    df["Time(seconds)"], marker="o", color="blue", label="Insertion Sort"
)

plt.title("Execution Time vs Number of Elements", fontsize=18, fontweight="bold")
plt.xlabel("Number of Elements (n)", fontsize=14)
plt.ylabel("Average Execution Time (seconds)", fontsize=14)

plt.grid(True, which="both", linestyle="--", alpha=0.4)
plt.legend()

plt.tight_layout()
plt.savefig("heap_sort_graph.png", dpi=300)
plt.show()
```
