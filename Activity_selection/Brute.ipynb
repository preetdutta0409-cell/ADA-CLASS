```python
import matplotlib.pyplot as plt

activities = [5, 10, 15, 20, 25, 30]

time_taken = [
    0.00000300,
    0.00000000,
    0.00200000,
    0.04500000,
    1.07600000,
    24.61300000
]

plt.plot(activities, time_taken, marker='o')

plt.xlabel("Number of Activities")
plt.ylabel("Time (seconds)")
plt.title("Brute Force Activity Selection - Time Complexity")

plt.grid(True)
plt.show()
```
